```python
from pyspark.sql import SparkSession
spark = SparkSession.builder \
    .appName("Spark SQL Assignment") \
    .getOrCreate()
```

```python
# @title Schema Set Up & Json DATA
from pyspark.sql.types import *

schema = StructType([
    StructField("provider_id", StringType()),
    StructField("provider_name", StringType()),
    StructField("snapshot_date", StringType()),

    StructField("regions_by_code", MapType(
        StringType(),
        StructType([
            StructField("region_name", StringType()),
            StructField("grid_operator", StringType()),

            StructField("substations", ArrayType(
                StructType([
                    StructField("substation_id", StringType()),
                    StructField("substation_name", StringType()),

                    StructField("feeder_lines", ArrayType(
                        StructType([
                            StructField("feeder_id", StringType()),
                            StructField("voltage_level", StringType()),

                            StructField("households", ArrayType(
                                StructType([
                                    StructField("household_id", StringType()),
                                    StructField("customer_type", StringType()),

                                    StructField("metering_branch", StructType([
                                        StructField("meters", ArrayType(
                                            StructType([
                                                StructField("meter_id", StringType()),
                                                StructField("meter_type", StringType()),

                                                StructField("readings", ArrayType(
                                                    StructType([
                                                        StructField("reading_time", StringType()),
                                                        StructField("kwh_used", DoubleType()),
                                                        StructField("peak_flag", BooleanType())
                                                    ])
                                                ))
                                            ])
                                        ))
                                    ])),

                                    StructField("billing_branch", StructType([
                                        StructField("bills", ArrayType(
                                            StructType([
                                                StructField("bill_id", StringType()),
                                                StructField("billing_month", StringType()),

                                                StructField("charges", ArrayType(
                                                    StructType([
                                                        StructField("charge_id", StringType()),
                                                        StructField("charge_type", StringType()),
                                                        StructField("charge_amount", DoubleType()),

                                                        StructField("taxes_by_type", MapType(
                                                            StringType(),
                                                            StructType([
                                                                StructField("tax_rate", DoubleType()),
                                                                StructField("tax_amount", DoubleType())
                                                            ])
                                                        ))
                                                    ])
                                                ))
                                            ])
                                        ))
                                    ]))
                                ])
                            ))
                        ])
                    ))
                ])
            ))
        ])
    ))
])


json_data = """
[
  {
    "provider_id": "EP1",
    "provider_name": "IberGrid Energy",
    "snapshot_date": "2026-01-01",
    "regions_by_code": {
      "MAD": {
        "region_name": "Madrid",
        "grid_operator": "Central Grid Operator",
        "substations": [
          {
            "substation_id": "SS1",
            "substation_name": "Madrid North Substation",
            "feeder_lines": [
              {
                "feeder_id": "F1",
                "voltage_level": "medium",
                "households": [
                  {
                    "household_id": "H1",
                    "customer_type": "residential",
                    "metering_branch": {
                      "meters": [
                        {
                          "meter_id": "M1",
                          "meter_type": "smart",
                          "readings": [
                            {
                              "reading_time": "2026-01-01T08:00:00",
                              "kwh_used": 2.4,
                              "peak_flag": false
                            },
                            {
                              "reading_time": "2026-01-01T19:00:00",
                              "kwh_used": 4.8,
                              "peak_flag": true
                            }
                          ]
                        }
                      ]
                    },
                    "billing_branch": {
                      "bills": [
                        {
                          "bill_id": "B1",
                          "billing_month": "2026-01",
                          "charges": [
                            {
                              "charge_id": "C1",
                              "charge_type": "energy_usage",
                              "charge_amount": 32.5,
                              "taxes_by_type": {
                                "VAT": {
                                  "tax_rate": 0.21,
                                  "tax_amount": 6.83
                                },
                                "GRID_FEE": {
                                  "tax_rate": 0.05,
                                  "tax_amount": 1.63
                                }
                              }
                            }
                          ]
                        }
                      ]
                    }
                  },
                  {
                    "household_id": "H2",
                    "customer_type": "residential",
                    "metering_branch": {
                      "meters": [
                        {
                          "meter_id": "M2",
                          "meter_type": "smart",
                          "readings": [
                            {
                              "reading_time": "2026-01-02T09:00:00",
                              "kwh_used": 3.1,
                              "peak_flag": false
                            }
                          ]
                        }
                      ]
                    },
                    "billing_branch": {
                      "bills": [
                        {
                          "bill_id": "B2",
                          "billing_month": "2026-01",
                          "charges": [
                            {
                              "charge_id": "C2",
                              "charge_type": "energy_usage",
                              "charge_amount": 28.0,
                              "taxes_by_type": {
                                "VAT": {
                                  "tax_rate": 0.21,
                                  "tax_amount": 5.88
                                },
                                "METER_RENTAL": {
                                  "tax_rate": 0.03,
                                  "tax_amount": 0.84
                                }
                              }
                            }
                          ]
                        }
                      ]
                    }
                  }
                ]
              }
            ]
          }
        ]
      }
    }
  }
]
"""

df_source = spark.read.schema(schema).json(spark.sparkContext.parallelize([json_data]))
```
